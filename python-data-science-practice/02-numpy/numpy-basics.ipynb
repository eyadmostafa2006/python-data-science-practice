## Part 3: Indexing and Slicing (1D, 2D, and 3D)

### Task 3.1: 1D Array Indexing & Slicing
Given the array: `arr_1d = np.array([15, 25, 35, 45, 55, 65])`
1. Access and print the first element.
2. Access and print the last element using negative indexing.
3. Slice and print elements from index `1` to `4` (inclusive of 1, exclusive of 4).

````python
arr_1d = np.array([15, 25, 35, 45, 55, 65])
print(arr_1d[0])    # first element
print(arr_1d[-1])   # last element
print(arr_1d[1:4])  # index 1 up to (not including) 4
````

### Task 3.2: 2D Array Indexing & Slicing
Given the matrix:
```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
```
1. Get the value `50` (row index 1, column index 1).
2. Get the entire second row (`[40, 50, 60]`).
3. Get the last column (`[30, 60, 90]`).
4. Slice the top-left 2x2 sub-array (`[[10, 20], [40, 50]]`).

````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print(matrix[1, 1])        # single value
print(matrix[1])           # second row
print(matrix[:, 2])        # last column
print(matrix[0:2, 0:2])    # top-left 2x2 block
````

### Task 3.3: 3D Array Indexing & Slicing
Given the 3D array:
```python
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
```
1. Print the first 2D block (matrix at index 0).
2. Access and print the value `7`.
3. Extract the first element from both blocks (i.e. `1` and `5`).

````python
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print(arr_3d[0])
print(arr_3d[1,1,0])
print(arr_3d[:,0,0])
````

## Part 4: Basic Math Operations

### Task 4.1: Element-wise Arithmetic
Given:
```python
a = np.array([2, 4, 6, 8])
b = np.array([1, 2, 3, 4])
```
1. Add `a` and `b`.
2. Subtract `b` from `a`.
3. Multiply `a` by `b`.
4. Divide `a` by `b`.

````python
a = np.array([2, 4, 6, 8])
b = np.array([1, 2, 3, 4])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
````

### Task 4.2: Scalar Operations

````python
a = np.array([10, 20, 30])
print(a + 5)
print(a * 2)
print(a ** 2)
````

### Task 4.3: Summary Functions
Given the matrix:
```python
data = np.array([
    [5, 10, 15],
    [20, 25, 30]
])
```
Find and print:
1. The sum of all elements.
2. The mean (average) of all elements.
3. The minimum and maximum values.
4. The sum of each column (`axis=0`).

````python
data = np.array([
    [5, 10, 15],
    [20, 25, 30]
])
print(data.sum())
print(data.mean())
print(data.max())
print(data.min())
print(data.sum(axis=0))  # axis=0 collapses the rows, so we get one sum per column
````
