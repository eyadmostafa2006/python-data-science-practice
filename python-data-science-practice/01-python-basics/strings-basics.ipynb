# Python Strings Basics

Short exercises on variables, types, and the most common string methods.

**Task 1**: Create a variable with a number and use the `type()` function to find its type.

```python
num = 20
print(type(num))
```

**Task 2:** Create a string and use `.capitalize()` to make the first character uppercase.

```python
Y = "i love python"
print(Y.capitalize())
```

**Task 3**: Create a string and print its length using the `len()` function.

```python
X = "I Go To Any Where"
len(X)
```

**Task 5**: Convert a string to uppercase and lowercase using `.upper()` and `.lower()`.

```python
txt = "OmAr"
print(txt.upper())
print(txt.lower())
```

**Task 6**: Find the index of a substring in a string using `.find()`.

```python
txt = "Python is a powerful programming language"
index = txt.find("powerful")
print(index)
```

**Task 7**: Replace a word in a string using `.replace()`.

```python
txt = "I like apples"
new_txt = txt.replace("apples", "oranges")
print(new_txt)
```

**Task 8**: Split a sentence into a list of words using `.split()`.

```python
txt = "This is a simple sentence"
words = txt.split()
print(words)
```

**Task 9**: Check if a string starts with or ends with a specific substring using `.startswith()` and `.endswith()`.

```python
txt = "Hello world!"
print(txt.startswith("Hello"))
print(txt.endswith("world!"))
```

**Task 10**: Remove whitespace from the beginning and end of a string using `.strip()`.

```python
txt = "   Hello world!   "
clean_txt = txt.strip()
print(clean_txt)
```

**Task 11**: Use string formatting to insert variables into a string.

```python
name = "Ahmed"
age = 25
txt = f"My name is {name} and I am {age} years old."
print(txt)
```

**Task 12**: Access the first and last character of a string using indexing.

```python
txt = "Python"
first_char = txt[0]
last_char = txt[-1]
print(first_char)
print(last_char)
```
