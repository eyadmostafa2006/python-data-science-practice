## Problem 3: Convert Minutes to Seconds
Given a number of minutes, print how many seconds that is.

Example: `3` minutes → `180` seconds

```python
minutes = 3
seconds = minutes * 60
print("Number of seconds:", seconds)
```

## Problem 4: Check Even or Odd
Given a number, print `"Even"` if it's even and `"Odd"` if it's odd.

```python
num = 8
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## Problem 5: Largest of Two Numbers
Given two numbers, print which one is larger.

```python
a = 12
b = 7
if a > b:
    print(a, "is larger")
elif b > a:
    print(b, "is larger")
else:
    print("Both numbers are equal")
```

## Problem 6: Simple Grade Checker

```python
score = 72
if score >= 50:
    print("Pass")
else:
    print("Fail")
```

## Problem 7: Print Numbers 1 to 20
Use a `for` loop to print all numbers from 1 to 20.

```python
for i in range(1, 21):
    print(i)
```

## Problem 8: Print Only Even Numbers
Use a `for` loop to print only the even numbers from 1 to 20.

```python
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

## Problem 9: Sum of Numbers 1 to N
Given `N = 10`, use a loop to calculate the sum of all numbers from 1 to N.

```python
N = 10
total = 0
for i in range(1, N + 1):
    total += i
print(total)
```

## Problem 10: Count Down From 10
Use a `while` loop to print numbers from 10 down to 1.

```python
n = 10
while n >= 1:
    print(n)
    n -= 1
```

## Problem 11: Multiplication Table
Given a number, print its multiplication table from 1 to 10 using a `for` loop.

Example: for `number = 3` print `3 x 1 = 3`, `3 x 2 = 6`, ... up to `3 x 10 = 30`

```python
number = 3
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

## Problem 12: Count Length of a List

## Problem 13: Sum of a List
Given a list of numbers, use a loop to calculate and print the sum of all its items (don't use the built-in `sum()`).

```python
numbers = [10, 20, 30, 40, 50]
total = 0
for num in numbers:
    total += num
print(total)
```

## Problem 14: Find a Number in a List
Given a list of numbers and a target number, check if the target number exists in the list and print `"Found"` or `"Not Found"`.

```python
numbers = [10, 20, 30, 40, 50]
target = 30
if target in numbers:
    print("Found")
else:
    print("Not Found")
```

## Problem 15: Print Each Letter of a Word

## Problem 16: Count a Specific Letter
Given a word, count how many times the letter `"a"` appears in it using a loop.

```python
word = "banana"
count = 0
for letter in word:
    if letter == "a":
        count += 1
print(count)
```

## Problem 17: Simple Greeting Function
Write a function called `greet` that takes a `name` and prints `"Hello, <name>!"`. Call the function with your name.

## Problem 18: Square a Number (Function)
Write a function called `square` that takes a number and returns its square (number × number). Then print the result of calling it with `7`.

```python
def square(number):
    return number * number

print(square(7))
```

## Problem 19: Is Positive, Negative, or Zero (Function)
Write a function called `check_number` that takes a number and returns `"Positive"`, `"Negative"`, or `"Zero"` depending on its value.

```python
def check_number(number):
    if number > 0:
        return "Positive"
    elif number < 0:
        return "Negative"
    else:
        return "Zero"

print(check_number(-5)) 
```

## Problem 20: Average of Three Numbers (Function)
Write a function called `average` that takes three numbers and returns their average.

```python
def average(a, b, c):
    return (a + b + c) / 3

print(average(4, 8, 6))
```
